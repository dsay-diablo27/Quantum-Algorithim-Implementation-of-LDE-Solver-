# Numerical Solution of the Burgers' Equation

## Problem Statement :

<span style="color: blue"> We aim to simulate the **Burgers' equation** on the domain \( x belongs to [0, 1] \) with the following conditions:</span>



 $$ \color{black}\frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} = \nu \frac{\partial^2 u}{\partial x^2} $$



- <span style="color: orange"> **Boundary Conditions:**  </span>
    - $ u(0, t) = U_L $ left boundary
    - $ u(1, t) = U_R $ right boundary &nbsp;  

- <span style="color: orange">**Initial Condition:** </span>
    - $ u(x, 0) = 1 $  for  $ x < 0.5 $, and $ u(x, 0) = 0 $ otherwise   &nbsp;   &nbsp;  &nbsp;


## The Cole-Hopf Transformation

<span style="color: orange">The Cole-Hopf transformation linearizes the Burgers' equation:</span>

1. <span style="color: orange">**Transformation:**</span>
      
      
      $$ \color{black} u(x,t) = -2\nu \frac{\partial}{\partial x} \ln \phi(x,t) $$
- where  $ \phi(x, t) $  satisfies the heat equation:
     
     $$ \color{black} \frac{\partial \phi}{\partial t} = \nu \frac{\partial^2 \phi}{\partial x^2} $$

2. <span style="color: orange">**Initial Condition :**</span>
    $$
     \color{black}\phi(x, 0) = \exp\left(-\frac{1}{2\nu} \int_0^x u_0(y) dy \right)
    $$

## The Numerical Solution: Crank-Nicolson Method

To simulate the time evolution numerically, we use the Crank-Nicolson method:

- **Discretization:**  
    The spatial domain is discretized into grid points.  
    The Crank-Nicolson scheme is applied to the heat equation for $ \phi $.

- **Time Stepping:**  
    At each time step, solve for $ \phi(x, t) $, then recover $ u(x, t) $ using the Cole-Hopf transformation.

- **Boundary Conditions:**  
    Apply the specified Dirichlet boundary conditions for $ u $ at $ x=0 $ and $ x=1 $.

## Summary of Steps

1. Define initial and boundary conditions for \( u \).
2. Transform initial condition to \( \phi(x, 0) \).
3. Numerically solve the heat equation for \( \phi(x, t) \) using Crank-Nicolson.
4. Recover \( u(x, t) \) from \( \phi(x, t) \) via Cole-Hopf.
5. Visualize the evolution of \( u(x, t) \) over time.
